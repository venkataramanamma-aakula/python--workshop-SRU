```python
# Break statement( if the condition is satisfied the loop will go to terminated)
k=int(input("enter value"))
i=1
while(i<=10):
  print (f"{k} * {i} = {k*i}")
  if(i==5):
    break
  i=i+1


```

```python
# continue statement(skip the current iteration after loops  print  )
k=int(input("enter value"))
i=0
while(i<=10):
  i=i+1
  if(i==5):
    continue
  print (f"{k} * {i} = {k*i}")




```
